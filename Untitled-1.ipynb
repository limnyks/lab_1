```python
import numpy as np 
import matplotlib.pyplot as plt
import math as m
```

```python
with open('banan.csv') as f:
    text = f.read().replace(',', '.')

data = np.loadtxt(text.splitlines(), delimiter=';')
data = data.T 
print(data.shape)

#дістаємо дані з файлу та транспонуємо матрицю для наступних перетворень
```

```python
plt.plot(data[1], data[0], color='yellow', label='Banana')
```

```python
#PART 1. Task 1
# Створюємо функції для кожного з перетворень, які будуть застосовані до даних
def stretch(x,a,b):
    stretch_matrix = np.array([[a,0],[0,b]])
    return np.dot(stretch_matrix, x)


def  shear(x,a,b):
    shear_matrix = np.array([[1,a],[b,1]])
    return np.dot(shear_matrix, x)


def  reflection(x,a,b):
    reflection_matrix = (1/(a**2+b**2)) * np.array([[a**2 - b**2,2*a*b],[2*a*b,b**2 - a**2]])
    return np.dot(reflection_matrix, x)

def  rotation(x,theta):
    rotation_matrix = np.array([[m.cos(theta),-m.sin(theta)],[m.sin(theta),m.cos(theta)]])
    return np.dot(rotation_matrix, x)


```

```python
shear_data = shear(data, 0.5, 0)
plt.plot(shear_data[1], shear_data[0], color='blue', label='Sheared Banana')
plt.plot(data[1], data[0], color='yellow', label='Original Banana')
```

```python
stretch_data = stretch(data, 3, 3)
plt.plot(stretch_data[1], stretch_data[0], color='red', label='Stretched Banana')
plt.plot(data[1], data[0], color='yellow', label='Banana')
plt.legend()
plt.show()
```

```python
reflection_data = reflection(data, 1, 1)
plt.plot(reflection_data[1], reflection_data[0], color='yellow', label='Banana')
plt.plot(data[1], data[0], color='blue', label='Original Banana')
plt.legend()
plt.show()
```

```python
rotation_data = rotation(data, m.pi/2)
plt.plot(rotation_data[1], rotation_data[0], color='green', label='Rotated Banana')
plt.plot(data[1], data[0], color='yellow', label='Original Banana')
plt.legend()
plt.show()
```

```python
def read_off(filename: str):
    with open(filename, 'r') as f:
        # Перевіряємо, чи перший рядок починається з OFF
        if 'OFF' != f.readline().strip():
            raise ValueError('Not a valid OFF header')

        # Зчитуємо кількість вершин, граней та ребер (третє значення часто ігнорується)
        n_verts, n_faces, _ = map(int, f.readline().strip().split())

        # Зчитуємо координати всіх вершин (x, y, z)
        verts = [list(map(float, f.readline().strip().split())) for _ in range(n_verts)]

        # Зчитуємо грані: перше число у рядку - кількість вершин грані (ігноруємо), далі індекси
        faces = [list(map(int, f.readline().strip().split()[1:])) for _ in range(n_faces)]

    # Повертаємо вершини у вигляді масиву NumPy та список граней
    f = np.array(verts)
    return f, faces


vertices, faces = read_off("airplane_0001.off")
print(vertices.shape)
print(faces[:3])
print(vertices)
```

```python
# Імпортуємо matplotlib для побудови 3D-графіків
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

# Функція для візуалізації OFF-моделі у вигляді сітки та точок
# (зручно бачити, як змінюється модель після трансформацій)
def plot_off(vertices, faces):
    fig = plt.figure(figsize=(8, 8))                   # створюємо вікно
    ax = fig.add_subplot(111, projection='3d')         # додаємо 3D координатну систему

    # Створюємо полігональну сітку з граней (faces) та додаємо її на графік
    mesh = Poly3DCollection([vertices[face] for face in faces],
                            alpha=0.3, edgecolor='k')  # прозорість 0.3, чорні ребра
    ax.add_collection3d(mesh)

    # Додаємо вершини як червоні точки
    ax.scatter(vertices[:, 0], vertices[:, 1], vertices[:, 2], s=0.1, c='b')

    # Підписуємо осі
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Z")

    # Автоматично масштабуємо сцену під модель
    ax.auto_scale_xyz(vertices[:, 0], vertices[:, 1], vertices[:, 2])

    plt.show()  # показуємо результат

plot_off(vertices, faces)
```

```python
def rotate_xy(vertice, angle):
    rotation_matrix = np.array([[np.cos(angle), -np.sin(angle), 0],[np.sin(angle),  np.cos(angle), 0],[0,0,1]])
    rotated_vertices_xy = np.dot(vertice, rotation_matrix.T)  
    
    return rotated_vertices_xy


def rotate_yz(vertice, angle):
    rotation_matrix = np.array([[1, 0, 0],[0,  np.cos(angle), np.sin(angle)],[0,-np.sin(angle), np.cos(angle)]])
    rotated_vertices_yz = np.dot(vertice, rotation_matrix.T)  
    
    return rotated_vertices_yz

def rotate_xz(vertice, angle):
    rotation_matrix = np.array([[np.cos(angle), 0, -np.sin(angle)],[0,  1, 0],[-np.sin(angle), 0, np.cos(angle)]])
    rotated_vertices_xz = np.dot(vertice, rotation_matrix.T)  
    
    return rotated_vertices_xz
```

```python

plot_off(rotate_xy(vertices, np.pi / 4), faces)
```

```python
plot_off(rotate_xz(vertices, np.pi / 4), faces)
```

```python
plot_off(rotate_yz(vertices, np.pi / 4), faces)
```

```python
first_run = rotate_yz(vertices, np.pi / 4)
first_run= rotate_xy(first_run, np.pi / 4)
plot_off(rotate_xz(first_run, np.pi / 4), faces)
```

```python
second_run = rotate_xz(vertices, np.pi / 4)
second_run= rotate_yz(second_run, np.pi / 4)
plot_off(rotate_xy(second_run, np.pi / 4), faces)
```
